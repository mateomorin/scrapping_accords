# Récupération des métadonnées

```python
client = LegifranceClient()
payload = {
  "recherche": {
    "filtres": [
      {
        "dates": {
          "start": "2020-03-01",
          "end": "2020-03-31"
        },
        "facette": "DATE_SIGNATURE"
      }
    ],
    "sort": "ID",
    "fromAdvancedRecherche": False,
    "pageSize": 100,
    "typePagination": "DEFAUT",
    "pageNumber": 1
  },
  "fond": "ACCO"
}
response = client.call_api("search", data=payload)
```

# Récupération des textes

```python
payload = {
  "id": "ACCOTEXT000038257107"
}

response = client.call_api("consult/acco", data=payload)
```

```python
example_data = response.json()["acco"]["data"]
```

```python
example_data
```

```python
from docx import Document
from io import BytesIO
import base64

binary_data = base64.b64decode(example_data)
data = BytesIO(binary_data)
document = Document(data)
```

```python
full_text = [paragraph.text for paragraph in document.paragraphs]
text = "\n".join(full_text)

text
```
